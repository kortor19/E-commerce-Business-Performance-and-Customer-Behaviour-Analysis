## Clean and prepare the data for analysis.

```python
# 1. Convert timestamp columns to datetime object
datetime_cols = ['created_at']
for df in [df_website_sessions, df_website_pageviews, df_products, df_orders_mav, df_order_items, df_order_item_refunds]:
    for col in datetime_cols:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')

# 2. Fill missing traffic UTM sources in df_website_sessions
# If utm_source is null but http_referer is present, it's typically organic search.
# If both are null, it is direct traffic.
resolved_sources = pd.Series(
    np.where(df_website_sessions['http_referer'].notnull(), 'organic_search', 'direct'),
    index=df_website_sessions.index
)

df_website_sessions['utm_source'] = df_website_sessions['utm_source'].fillna(resolved_sources)
df_website_sessions['utm_campaign'] = df_website_sessions['utm_campaign'].fillna('organic_or_direct')
df_website_sessions['utm_content'] = df_website_sessions['utm_content'].fillna('organic_or_direct')
df_website_sessions['http_referer'] = df_website_sessions['http_referer'].fillna('direct_type')

# Display verification of changes
print("--- Website Sessions: Missing Values Cleaned ---")
print(df_website_sessions.isnull().sum())

print("\n--- Verification of Datatypes ---")
print(f"Sessions 'created_at' type: {df_website_sessions['created_at'].dtype}")
print(f"Orders 'created_at' type: {df_orders_mav['created_at'].dtype}")
```

###- All columns named created_at across all six tables (df_website_sessions, df_website_pageviews, df_products, df_orders_mav, df_order_items, and df_order_item_refunds) have been successfully converted from raw strings (object) into proper pandas datetime timestamps
### - This conversion guarantees that all future calculations based on time (such as calculating session-to-order durations, quarterly performance trends, and monthly return cycles) will be precise and robust.

### Initially, df_website_sessions contained significant gaps in its marketing attribution fields—specifically 83,328 missing UTM parameter entries and 39,917 missing referrer records. To preserve this traffic data for analysis, these null values were systematically imputed using business-logic fallbacks (classifying sessions with a referrer as organic search, and those without either parameter as direct traffic)

# Task 2: Exploratory data analysis

## How does revenue change over time?

```python
# how does revenue change over time

# 1. Extract chronological monthly periods from the transaction datetimes
df_orders_mav['year_month'] = df_orders_mav['created_at'].dt.to_period('M')

# 2. Group by month and calculate sum of revenue
monthly_revenue = df_orders_mav.groupby('year_month')['price_usd'].sum().reset_index()
monthly_revenue.columns = ['Month', 'Revenue_USD']

# 3. Calculate Month-over-Month growth rate
monthly_revenue['MoM_Growth_%'] = monthly_revenue['Revenue_USD'].pct_change() * 100

# Keep a formatted version for clean tabular display
formatted_revenue = monthly_revenue.copy()
formatted_revenue['Revenue_USD'] = formatted_revenue['Revenue_USD'].map('${:,.2f}'.format)
formatted_revenue['MoM_Growth_%'] = formatted_revenue['MoM_Growth_%'].map(lambda x: f"{x:+.2f}%" if not pd.isna(x) else "N/A")

print("--- Monthly Revenue and MoM Growth Summary ---")
display(formatted_revenue)

# 4. Plot the revenue trend over time
# Convert Month back to timestamps for correct plotting scales
monthly_revenue['Month_Timestamp'] = monthly_revenue['Month'].dt.to_timestamp()

plt.figure(figsize=(14, 6))
sns.lineplot(data=monthly_revenue, x='Month_Timestamp', y='Revenue_USD', marker='o', color='#4C72B0', linewidth=2.5)
plt.title('Monthly Sales Revenue Trend', fontsize=16, fontweight='bold', pad=15)
plt.xlabel('Date (Monthly Intervals)', fontsize=12)
plt.ylabel('Total Sales (USD)', fontsize=12)
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: "${:,.0f}".format(x)))
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```


### Revenue Growth Summary: Maven Fuzzy Factory experienced significant revenue growth from its launch in March 2012, with monthly sales increasing from $2,999.40 to $30,893.82 by November 2012 (+66.58% MoM). Revenue continued to rise, reaching $58,262.60 in December 2013 and peaking at $144,823.02 in December 2014, representing approximately 148% YoY growth.

### There is a consistent, highly predictable surge in revenue during the late-holiday quarters (November & December): In 2012, November saw +66.58% MoM growth. In 2013, November and December grew by +21.93% and +24.94% respectively. In 2014, November and December rose by +23.35% and +13.00% respectively. This holiday surge is immediately followed by a post-holiday decline in January of each year (e.g., -18.12% in Dec 2012 / -21.07% in Jan 2013, and -8.71% in Jan 2015).

### In March 2015, the revenue shows an abrupt decrease of -38.90%, falling to $78,951.07 from $78,951.07 from $129,212.94 in February.

### **Important Note**:
 ### This drop is almost certainly due to incomplete data for March 2015 (a truncated reporting period) rather than an actual business collapse. The timestamps in this dataset likely cut off mid-month in March 2015.

## Which products generate the highest sales?

```python
# Which products generate the highest sales?

# 1. Merge order items with product metadata
df_product_sales = df_order_items.merge(df_products, on='product_id', how='left')

# 2. Aggregate sales metrics at the product level
product_summary = df_product_sales.groupby(['product_id', 'product_name']).agg(
    units_sold=('order_item_id', 'count'),
    total_revenue_usd=('price_usd', 'sum'),
    total_cogs_usd=('cogs_usd', 'sum')
).reset_index()

# 3. Calculate gross profit and margins
product_summary['gross_profit_usd'] = product_summary['total_revenue_usd'] - product_summary['total_cogs_usd']
product_summary['gross_margin_%'] = (product_summary['gross_profit_usd'] / product_summary['total_revenue_usd']) * 100

# Sort by total revenue descending
product_summary = product_summary.sort_values(by='total_revenue_usd', ascending=False)

# Formatted DataFrame for display
formatted_summary = product_summary.copy()
formatted_summary['total_revenue_usd'] = formatted_summary['total_revenue_usd'].map('${:,.2f}'.format)
formatted_summary['total_cogs_usd'] = formatted_summary['total_cogs_usd'].map('${:,.2f}'.format)
formatted_summary['gross_profit_usd'] = formatted_summary['gross_profit_usd'].map('${:,.2f}'.format)
formatted_summary['gross_margin_%'] = formatted_summary['gross_margin_%'].map('{:.2f}%'.format)
formatted_summary['units_sold'] = formatted_summary['units_sold'].map('{:,}'.format)

print("--- Product Sales Performance Summary ---")
display(formatted_summary)

# 4. Plot the product revenue
plt.figure(figsize=(10, 5))
sns.barplot(data=product_summary, x='total_revenue_usd', y='product_name', palette='Blues_r')
plt.title('Gross Revenue generated by Product', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Gross Revenue (USD)', fontsize=11)
plt.ylabel('Product Name', fontsize=11)
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: "${:,.0f}".format(x)))
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### **Top Revenue Generator**: **The Original Mr. Fuzzy** (Product 1) is the clear best-seller, driving $1,211,057.74 in sales (over 77% of total revenue) with 24,226 units sold.

### Runner-Up: The Forever Love Bear (Product 2) is second, generating **$347,702.04** from 5,796 units.

### Highest Profit Margins: While The Birthday Sugar Panda (Product 3) and The Hudson River Mini bear (Product 4) have lower unit sales (5,000 each), they yield significantly higher profit margins (**68.4%**) than the best-sellers (~61–62%).

## What are the patterns in customer purchasing behaviour?

```python
# What are the patterns in customer purchasing behaviour?
# 1. Repeat Session Analysis
session_summary = df_website_sessions.groupby('is_repeat_session').agg(
    total_sessions=('website_session_id', 'count')
).reset_index()
session_summary['percentage'] = (session_summary['total_sessions'] / session_summary['total_sessions'].sum()) * 100

# Merge sessions with orders to see conversion by session type
df_sessions_orders = df_website_sessions.merge(
    df_orders_mav[['website_session_id', 'order_id', 'price_usd', 'items_purchased']],
    on='website_session_id',
    how='left'
)

# Flag converted sessions
df_sessions_orders['is_converted'] = df_sessions_orders['order_id'].notnull().astype(int)

conversion_by_session_type = df_sessions_orders.groupby('is_repeat_session').agg(
    total_sessions=('website_session_id', 'count'),
    total_orders=('order_id', 'count'),
    avg_revenue_per_order=('price_usd', 'mean'),
    avg_items_per_order=('items_purchased', 'mean')
).reset_index()

conversion_by_session_type['conversion_rate_%'] = (conversion_by_session_type['total_orders'] / conversion_by_session_type['total_sessions']) * 100

print("--- Purchasing Behavior: New vs Repeat Sessions ---")
display(conversion_by_session_type)

# 2. Distribution of items purchased per order
items_dist = df_orders_mav['items_purchased'].value_counts(normalize=True).reset_index()
items_dist.columns = ['Items_Purchased', 'Proportion_of_Orders']
items_dist['Proportion_of_Orders_%'] = items_dist['Proportion_of_Orders'] * 100

print("\n--- Number of Items Purchased per Order ---")
display(items_dist)
```

```python
# 3. Visualize Conversion and Revenue by Session Type
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Conversion rate plot
sns.barplot(
    data=conversion_by_session_type,
    x='is_repeat_session',
    y='conversion_rate_%',
    ax=axes[0],
    palette='Blues'
)
axes[0].set_title('Conversion Rate: New vs. Repeat Sessions', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Is Repeat Session (0 = New, 1 = Repeat)')
axes[0].set_ylabel('Conversion Rate (%)')
axes[0].set_xticklabels(['New Session', 'Repeat Session'])

# Average revenue plot
sns.barplot(
    data=conversion_by_session_type,
    x='is_repeat_session',
    y='avg_revenue_per_order',
    ax=axes[1],
    palette='Greens'
)
axes[1].set_title('Average Order Value (AOV): New vs. Repeat', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Session Type')
axes[1].set_ylabel('AOV (USD)')
axes[1].set_xticklabels(['New Session', 'Repeat Session'])

plt.tight_layout()
plt.show()
```

### **Customer Behaviour and Purchasing Patterns Summary**
#### Repeat visitors demonstrate a higher purchase conversion rate (7.83%) than first-time visitors (6.64%), indicating stronger purchasing intent among returning users. However, their average order values remain relatively similar ($59.86 for new sessions and $60.54 for repeat sessions), with comparable basket sizes of 1.23–1.25 items per order.

#### Regarding purchasing patterns, 76.13% of orders contain a single item, while 23.87% contain two items. No orders containing three or more items were recorded, suggesting opportunities to increase basket size through cross-selling, product bundling, and promotional incentives.

## How does website traffic differ between mobile and desktop users?

```python
# How does website traffic differ between mobile and desktop users?

# 1. Merge website sessions with order data to identify converted sessions by device type
df_sessions_orders_device = df_website_sessions.merge(
    df_orders_mav[['website_session_id', 'order_id', 'price_usd']],
    on='website_session_id',
    how='left'
)

# 2. Aggregate sessions, orders, and sales by device type
device_summary = df_sessions_orders_device.groupby('device_type').agg(
    total_sessions=('website_session_id', 'count'),
    total_orders=('order_id', 'count'),
    total_revenue=('price_usd', 'sum')
).reset_index()

# 3. Calculate key performance indicators (KPIs)
device_summary['traffic_share_%'] = (device_summary['total_sessions'] / device_summary['total_sessions'].sum()) * 100
device_summary['conversion_rate_%'] = (device_summary['total_orders'] / device_summary['total_sessions']) * 100
device_summary['average_order_value_usd'] = device_summary['total_revenue'] / device_summary['total_orders']

print("--- Traffic and Sales Breakdown by Device Type ---")
display(device_summary)

# 4. Visualize the comparison
fig, axes = plt.subplots(1, 2, figsize=(15, 5))

# Traffic Share Chart
axes[0].pie(
    device_summary['total_sessions'],
    labels=device_summary['device_type'].str.capitalize(),
    autopct='%1.1f%%',
    startangle=90,
    colors=['#5B92E5', '#A2C4C9']
)
axes[0].set_title('Website Traffic Share by Device Type', fontsize=12, fontweight='bold')

# Conversion Rate Chart
sns.barplot(
    data=device_summary,
    x='device_type',
    y='conversion_rate_%',
    ax=axes[1],
    palette='Blues_r'
)
axes[1].set_title('Conversion Rate by Device Type', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Device Type')
axes[1].set_ylabel('Conversion Rate (%)')
axes[1].set_xticklabels(['Desktop', 'Mobile'])

plt.tight_layout()
plt.show()
```

#### Desktop users are the primary revenue drivers, generating 69.16% of website traffic (327,027 sessions) and achieving an 8.50% conversion rate, resulting in 27,805 orders. Their strong traffic volume and conversion efficiency contribute $1.67 million in revenue, representing approximately 86% of total sales.

#### Mobile users account for 30.84% of website traffic (145,844 sessions) but have a low conversion rate of 3.09%, nearly one-third of the desktop rate. Consequently, mobile generates only 4,508 orders and $272,751.91 in revenue, highlighting a significant opportunity to improve the mobile shopping experience and conversion performance.

```python
!pip install -q reportlab

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Image, Table, TableStyle, PageBreak
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors

# Ensure all raw datasets are loaded to avoid NameErrors in separate runs
df_website_sessions = pd.read_csv('/content/website_sessions.csv')
df_products = pd.read_csv('/content/products.csv')
df_orders_mav = pd.read_csv('/content/orders_mav.csv')
df_order_items = pd.read_csv('/content/order_items.csv')
df_order_item_refunds = pd.read_csv('/content/order_item_refunds.csv')

# Ensure all dependencies are pre-calculated
# 1. Datetime conversions and basic revenue preparation
df_orders_mav['created_at'] = pd.to_datetime(df_orders_mav['created_at'], errors='coerce')
df_orders_mav['year_month'] = df_orders_mav['created_at'].dt.to_period('M')
monthly_revenue = df_orders_mav.groupby('year_month')['price_usd'].sum().reset_index()
monthly_revenue.columns = ['Month', 'Revenue_USD']
monthly_revenue['Month_Timestamp'] = monthly_revenue['Month'].dt.to_timestamp()

# 2. Product sales summary
df_product_sales = df_order_items.merge(df_products, on='product_id', how='left')
product_summary = df_product_sales.groupby(['product_id', 'product_name']).agg(
    units_sold=('order_item_id', 'count'),
    total_revenue_usd=('price_usd', 'sum'),
    total_cogs_usd=('cogs_usd', 'sum')
).reset_index()

# 3. Device performance metrics
df_sessions_orders_device = df_website_sessions.merge(
    df_orders_mav[['website_session_id', 'order_id', 'price_usd']],
    on='website_session_id',
    how='left'
)
device_summary = df_sessions_orders_device.groupby('device_type').agg(
    total_sessions=('website_session_id', 'count'),
    total_orders=('order_id', 'count'),
    total_revenue=('price_usd', 'sum')
).reset_index()
device_summary['conversion_rate_%'] = (device_summary['total_orders'] / device_summary['total_sessions']) * 100

# 4. Product refund summary
df_refund_analysis = (
    df_order_items
    .merge(df_order_item_refunds[['order_item_id', 'refund_amount_usd']], on='order_item_id', how='left')
    .merge(df_products[['product_id', 'product_name']], on='product_id', how='left')
)
refund_summary = df_refund_analysis.groupby(['product_id', 'product_name']).agg(
    units_sold=('order_item_id', 'count'),
    units_refunded=('refund_amount_usd', 'count'),
    total_refunded_usd=('refund_amount_usd', 'sum')
).reset_index()
refund_summary['refund_rate_%'] = (refund_summary['units_refunded'] / refund_summary['units_sold']) * 100

# Create directory for saving the plots
os.makedirs('/tmp/eda_plots', exist_ok=True)

# Plot 1: Monthly Revenue
plt.figure(figsize=(8, 3.5))
sns.lineplot(data=monthly_revenue, x='Month_Timestamp', y='Revenue_USD', marker='o', color='#1a5f7a', linewidth=2)
plt.title('Monthly Sales Revenue Trend', fontsize=11, fontweight='bold')
plt.xlabel('Timeline', fontsize=8)
plt.ylabel('Sales ($)', fontsize=8)
plt.tick_params(labelsize=8)
plt.gca().yaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"${x:,.0f}"))
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('/tmp/eda_plots/plot1.png', dpi=200)
plt.close()

# Plot 2: Product Gross Revenue
plt.figure(figsize=(8, 3.5))
sns.barplot(data=product_summary, x='total_revenue_usd', y='product_name', palette='Blues_r')
plt.title('Gross Revenue generated by Product', fontsize=11, fontweight='bold')
plt.xlabel('Revenue (USD)', fontsize=8)
plt.ylabel('')
plt.tick_params(labelsize=8)
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"${x:,.0f}"))
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('/tmp/eda_plots/plot2.png', dpi=200)
plt.close()

# Plot 3: Device Conversion
fig, axes = plt.subplots(1, 2, figsize=(8, 3.5))
axes[0].pie(device_summary['total_sessions'], labels=device_summary['device_type'].str.capitalize(), autopct='%1.1f%%', startangle=90, colors=['#5B92E5', '#A2C4C9'], textprops={'fontsize': 8})
axes[0].set_title('Traffic Share by Device', fontsize=10, fontweight='bold')
sns.barplot(data=device_summary, x='device_type', y='conversion_rate_%', ax=axes[1], palette='Blues_r')
axes[1].set_title('Conversion Rate by Device', fontsize=10, fontweight='bold')
axes[1].set_xlabel('')
axes[1].set_ylabel('Conversion %', fontsize=8)
axes[1].tick_params(labelsize=8)
plt.tight_layout()
plt.savefig('/tmp/eda_plots/plot3.png', dpi=200)
plt.close()

# Plot 4: Product Refund Rates
plt.figure(figsize=(8, 3.5))
sns.barplot(data=refund_summary, x='refund_rate_%', y='product_name', palette='Reds_r')
plt.title('Refund Rate (%) by Product', fontsize=11, fontweight='bold')
plt.xlabel('Refund Rate (%)', fontsize=8)
plt.ylabel('')
plt.tick_params(labelsize=8)
plt.gca().xaxis.set_major_formatter(plt.FuncFormatter(lambda x, loc: f"{x:.1f}%"))
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.savefig('/tmp/eda_plots/plot4.png', dpi=200)
plt.close()

print("Temporary report charts successfully created and saved!")
```

```python
# 2. Build the PDF Document using ReportLab
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Image, Table, TableStyle, PageBreak
from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
from reportlab.lib import colors

pdf_path = "/content/Maven_Fuzzy_Factory_EDA_Report.pdf"
doc = SimpleDocTemplate(pdf_path, pagesize=letter, rightMargin=40, leftMargin=40, topMargin=40, bottomMargin=40)
story = []

styles = getSampleStyleSheet()

# Custom Styles
title_style = ParagraphStyle(
    'ReportTitle', parent=styles['Heading1'],
    fontName='Helvetica-Bold', fontSize=24, textColor=colors.HexColor('#1a5f7a'),
    spaceAfter=15, alignment=1
)
subtitle_style = ParagraphStyle(
    'ReportSubtitle', parent=styles['Normal'],
    fontName='Helvetica-Oblique', fontSize=12, textColor=colors.HexColor('#555555'),
    spaceAfter=30, alignment=1
)
h1_style = ParagraphStyle(
    'SectionHeading', parent=styles['Heading2'],
    fontName='Helvetica-Bold', fontSize=16, textColor=colors.HexColor('#1a5f7a'),
    spaceBefore=15, spaceAfter=10
)
h2_style = ParagraphStyle(
    'SubSectionHeading', parent=styles['Heading3'],
    fontName='Helvetica-Bold', fontSize=12, textColor=colors.HexColor('#222222'),
    spaceBefore=8, spaceAfter=6
)
body_style = ParagraphStyle(
    'ReportBody', parent=styles['BodyText'],
    fontName='Helvetica', fontSize=10, leading=14, textColor=colors.HexColor('#333333'),
    spaceAfter=10
)
bullet_style = ParagraphStyle(
    'ReportBullet', parent=styles['Normal'],
    fontName='Helvetica', fontSize=10, leading=14, textColor=colors.HexColor('#333333'),
    leftIndent=20, firstLineIndent=-10, spaceAfter=6
)

# --- PAGE 1: TITLE & EXECUTIVE SUMMARY ---
story.append(Paragraph("Maven Fuzzy Factory", title_style))
story.append(Paragraph("E-commerce Performance & Customer Behaviour Analysis Report", subtitle_style))
story.append(Spacer(1, 15))

story.append(Paragraph("Executive Summary", h1_style))
story.append(Paragraph(
    "This report provides an in-depth analysis of Maven Fuzzy Factory's website performance, customer interaction dynamics, marketing channel attribution, and order transaction streams from launch in March 2012 to early 2015.",
    body_style
))

# Summary Key metrics table
summary_data = [
    ['Metric', 'Value', 'Primary Insight'],
    ['Total Website Traffic', '472,871 Sessions', 'Strong traffic growth driven primarily by paid search channels.'],
    ['Total Orders Registered', '32,313 Transactions', 'Reflects an overall session-to-order conversion rate of 6.83%.'],
    ['Total Revenue (USD)', '$1,958,414.43', 'Driven largely by our core bestseller product.'],
    ['Average Order Value', '$59.99', 'Consistent basket size of ~1.24 items per order.'],
    ['Average Return Rate', '4.32% (Line Items)', 'Acceptable, but significant variance observed across catalog.']
]
t_summary = Table(summary_data, colWidths=[130, 110, 290])
t_summary.setStyle(TableStyle([
    ('BACKGROUND', (0, 0), (-1, 0), colors.HexColor('#1a5f7a')),
    ('TEXTCOLOR', (0, 0), (-1, 0), colors.whitesmoke),
    ('ALIGN', (0, 0), (-1, -1), 'LEFT'),
    ('FONTNAME', (0, 0), (-1, 0), 'Helvetica-Bold'),
    ('FONTSIZE', (0, 0), (-1, -1), 9),
    ('BOTTOMPADDING', (0, 0), (-1, 0), 6),
    ('BACKGROUND', (0, 1), (-1, -1), colors.HexColor('#f9f9f9')),
    ('GRID', (0, 0), (-1, -1), 0.5, colors.HexColor('#dddddd')),
    ('VALIGN', (0, 0), (-1, -1), 'TOP'),
]))
story.append(t_summary)
story.append(Spacer(1, 20))

story.append(Paragraph("Strategic Actions & Recommendations:", h2_style))
story.append(Paragraph("• <b>Optimize Mobile Journey:</b> Mobile traffic represents 30.8% of visits but converts at only 3.09% (vs. 8.50% on Desktop). Checkout flow restructuring is critical.", bullet_style))
story.append(Paragraph("• <b>Product Revenue De-risking:</b> The Original Mr. Fuzzy commands 77% of revenue. Bundling with newer high-margin assets will diversify commercial reliance.", bullet_style))
story.append(Paragraph("• <b>Resolve Refund Discrepancies:</b> The Birthday Sugar Panda exhibits a disproportionate 6.04% return rate. Inspect suppliers and description parameters immediately.", bullet_style))
story.append(Paragraph("• <b>Scale Organic Inbound Channels:</b> Over 71% of traffic depends on paid engines. Structured investment in SEO and loyalty retargeting can reduce customer acquisition costs (CAC).", bullet_style))

story.append(PageBreak())

# --- PAGE 2: SALES REVENUE & PRODUCT METRICS ---
story.append(Paragraph("1. Financial & Product Performance Analysis", h1_style))
story.append(Paragraph(
    "Monthly revenue trends depict strong seasonality with predictable peaks during late-holiday quarters (November/December), followed by a regular post-holiday contraction in January.",
    body_style
))

# Insert Plot 1
story.append(Image('/tmp/eda_plots/plot1.png', width=500, height=218))
story.append(Spacer(1, 15))

story.append(Paragraph("Product Catalog Revenue Contribution", h2_style))
story.append(Paragraph(
    "The catalog relies significantly on the flagship product, 'The Original Mr. Fuzzy'. However, the newer additions ('The Birthday Sugar Panda' and 'The Hudson River Mini Bear') maintain stronger product margin profiles (68.4%) than the leader (~61.5%).",
    body_style
))

# Insert Plot 2
story.append(Image('/tmp/eda_plots/plot2.png', width=500, height=218))

story.append(PageBreak())

# --- PAGE 3: DEVICE METRICS & REFUND PERFORMANCE ---
story.append(Paragraph("2. Behavioral Dynamics & Product Refunds", h1_style))
story.append(Paragraph(
    "While desktop users remain the financial engine of the organization (generating 86% of revenue), mobile conversion optimizations present the largest untapped area of growth.",
    body_style
))

# Insert Plot 3
story.append(Image('/tmp/eda_plots/plot3.png', width=500, height=218))
story.append(Spacer(1, 15))

story.append(Paragraph("Refund Activity by Product Catalog", h2_style))
story.append(Paragraph(
    "Refund analysis shows that sales volume heavily influences the raw number of refunds. However, looking at the rate of refund requests tells a different story: 'The Birthday Sugar Panda' has a significantly higher rate of returns compared to all other products.",
    body_style
))

# Insert Plot 4
story.append(Image('/tmp/eda_plots/plot4.png', width=500, height=218))

# Build document
doc.build(story)
print(f"Executive EDA Report successfully compiled and saved to: {pdf_path}")
```
